# Result charts

This notebook reads `results/final/` and the Locust CSVs. If the files are missing it stops. It does not draw placeholder charts.

```python
from pathlib import Path
import pandas as pd

final = Path("results/final")
required = [final / "quality.csv", final / "serving.csv", final / "cost.csv"]
missing = [str(path) for path in required if not path.exists()]
if missing:
    raise SystemExit(
        "No measured CSVs yet. Run the eval and load test in RUNBOOK.md, "
        "write results/final/*.csv from those outputs, then re-run. Missing: "
        + ", ".join(missing)
    )
quality = pd.read_csv(final / "quality.csv")
serving = pd.read_csv(final / "serving.csv")
cost = pd.read_csv(final / "cost.csv")
quality
```

```python
import matplotlib.pyplot as plt

# Expect Locust CSVs named results/loadtest-<users>_stats.csv after the GPU run.
rows = []
for path in sorted(Path("results").glob("loadtest-*_stats.csv")):
    frame = pd.read_csv(path)
    label = path.name.split("_stats")[0]
    rows.append((label, frame))
if not rows:
    raise SystemExit("No Locust CSVs under results/. The concurrency charts need a real load test.")

# Adapt the column names to the Locust version you ran, then save:
# fig.savefig("results/final/throughput_latency.png", dpi=140)
print("Loaded", len(rows), "Locust files. Plot them once you have confirmed the columns.")
```
